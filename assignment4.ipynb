# Visualization line Chart


# monthly  student enrollments

```python
monthly_enrollment = (
    df.groupby(df['EnrollmentDate'].dt.to_period('M'))
      .size()
      .reset_index(name='Total_Enrollments')
)

monthly_enrollment['MonthYear'] = (
    monthly_enrollment['EnrollmentDate']
    .dt.strftime('%b %Y')
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_enrollment['MonthYear'],
    monthly_enrollment['Total_Enrollments'],
    marker='o',
    linewidth=2
)

plt.title('Monthly Student Enrollments')
plt.xlabel('Month')
plt.ylabel('Total Enrollments')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# Monthly Revenue Trend

```python
monthly_revenue = (
    df.groupby(df['EnrollmentDate'].dt.to_period('M'))['Revenue']
      .sum()
      .reset_index()
)

monthly_revenue['MonthYear'] = (
    monthly_revenue['EnrollmentDate']
    .dt.strftime('%b %Y')
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_revenue['MonthYear'],
    monthly_revenue['Revenue'],
    marker='o',
    linewidth=2
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

# Average course progress by month

```python
monthly_progress = (
    df.groupby(df['EnrollmentDate'].dt.to_period('M'))['CourseProgress']
      .mean()
      .reset_index()
)
monthly_progress['MonthYear'] = (
    monthly_progress['EnrollmentDate']
    .dt.strftime('%b %Y')
)
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_progress['MonthYear'],
    monthly_progress['CourseProgress'],
    marker='o',
    linewidth=2
)

plt.title('Average Course Progress by Month')
plt.xlabel('Month')
plt.ylabel('Average Course Progress (%)')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# Horizontal bar graph

# Top 10 Courses by Enrollment

```python
top_courses = (
    df.groupby('Course')['StudentID']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='Total_Students')
)

import numpy as np
plt.figure(figsize=(10, 5))

colors = plt.cm.plasma(np.linspace(0.2, 0.8, len(top_courses)))

bars = plt.barh(
    top_courses['Course'],
    top_courses['Total_Students'],
    color=colors,        
    edgecolor='black'
)

plt.title('Top Courses by Student Enrollment')
plt.xlabel('Number of Students')
plt.ylabel('Course')

plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.bar_label(bars, padding=3)
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

#  Pie charts

# Student Distribution by Course

```python
course_distribution = (
    df.groupby('Course')['StudentID']
      .count()
)
plt.figure(figsize=(6,6))

explode = [0.08 if x == course_distribution.max() else 0
           for x in course_distribution]

colors = ['#4E79A7','#F28E2B','#59A14F',
          '#E15759','#76B7B2','#EDC948']

plt.pie(
    course_distribution,
    labels=course_distribution.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=explode,
    shadow=True,
    colors=colors,
    wedgeprops={'edgecolor':'white','linewidth':1}
)

plt.title('Student Distribution by Course', fontsize=14, fontweight='bold')

plt.axis('equal')

plt.tight_layout()
plt.show()
```
