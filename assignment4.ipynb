# Visualization line Chart


# monthly  student enrollments

```python
monthly_enrollment = (
    df.groupby(df['EnrollmentDate'].dt.to_period('M'))
      .size()
      .reset_index(name='Total_Enrollments')
)

monthly_enrollment['MonthYear'] = (
    monthly_enrollment['EnrollmentDate']
    .dt.strftime('%b %Y')
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))


plt.plot(
    monthly_enrollment['MonthYear'],
    monthly_enrollment['Total_Enrollments'],
    marker='o',
    linewidth=2,
    color='royalblue'
)

for x, y in zip(monthly_enrollment['MonthYear'], monthly_enrollment['Total_Enrollments']):
    plt.text(
        x, y + (max(monthly_enrollment['Total_Enrollments']) * 0.02), 
        f'{y:,}',                                                     
        ha='center',                                               
        va='bottom',                                                 
        fontsize=9,
        fontweight='bold'
    )

plt.title('Monthly Student Enrollments', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Total Enrollments')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.ylim(top=max(monthly_enrollment['Total_Enrollments']) * 1.1)

plt.tight_layout()
plt.show()
```

# Monthly Revenue Trend

```python
monthly_revenue = (
    df.groupby(df['EnrollmentDate'].dt.to_period('M'))['Revenue']
      .sum()
      .reset_index()
)

monthly_revenue['MonthYear'] = (
    monthly_revenue['EnrollmentDate']
    .dt.strftime('%b %Y')
)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_revenue['MonthYear'],
    monthly_revenue['Revenue'],
    marker='o',
    linewidth=2,
    color='royalblue'
)

max_val = monthly_revenue['Revenue'].max()

for x, y in zip(monthly_revenue['MonthYear'], monthly_revenue['Revenue']):

    label = f'Rs. {y/1000:.0f}K' if y < 1e6 else f'Rs. {y/1e6:.1f}M'
    
    plt.text(
        x, y + (max_val * 0.02),
        label,
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )
plt.title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)


plt.ylim(top=max_val * 1.12)

plt.tight_layout()
plt.show()
```

# Average course progress by month

# Histogram

#  Course Fee Distribution

```python
import seaborn as sns

plt.figure(figsize=(8, 5))

ax = sns.histplot(
    df['CourseFee'],
    bins=10,
    color='royalblue',
    edgecolor='black',
    kde=True  
)

# --- Add count labels on top of each bin ---
for p in ax.patches:
    height = p.get_height()
    if height > 0:  # Skip empty bins
        ax.annotate(
            f'{int(height)}', 
            (p.get_x() + p.get_width() / 2., height), 
            ha='center', 
            va='bottom', 
            xytext=(0, 3), 
            textcoords='offset points',
            fontweight='bold',
            fontsize=9
        )

plt.title('Distribution of Course Fee', fontsize=14, fontweight='bold')
plt.xlabel('Course Fee')
plt.ylabel('Frequency')

plt.grid(axis='y', linestyle='--', alpha=0.5)

# Add headroom so top labels don't get cut off
plt.ylim(top=ax.get_ylim()[1] * 1.1)

plt.tight_layout()
plt.show()
```
